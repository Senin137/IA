1. Observando la gráfica temporal: ¿La serie de tiempo es estacionaria
     (su media y varianza son constantes)? ¿Por qué esto es un factor
     crítico antes de entrenar un modelo predictivo?

La serie de tiempo analizada no es estacionaria debido a que su media aumenta linealmente por la tendencia y su varianza se altera drásticamente con las anomalías inyectadas, lo cual representa un riesgo crítico para el modelado predictivo. En ingeniería de datos, entrenar con señales no estacionarias suele provocar regresiones espurias (correlaciones falsas que el modelo cree reales), inestabilidad numérica en los gradientes por valores fuera de rango y una nula capacidad de generalización, ya que los modelos de aprendizaje automático asumen que las reglas estadísticas del pasado se mantendrán constantes; si la distribución de los datos cambia continuamente (drift), el modelo se vuelve obsoleto al instante al no poder procesar valores que nunca vio durante su entrenamiento.

3. ¿Qué riesgos corremos como ingenieros si saltamos esta fase de
     análisis y conectamos directamente estos datos crudos a una Red
     Neuronal?

Saltar el EDA e introducir datos crudos a una Red Neuronal condena al proyecto al principio de "Garbage In, Garbage Out", donde el modelo aprende errores de hardware o interferencias electromagnéticas como si fueran patrones físicos reales, derivando en predicciones falsas y poco fiables. Técnicamente, la presencia de valores atípicos extremos o nulos provoca inestabilidad numérica, como gradientes explosivos que rompen el proceso de optimización al generar pesos matemáticamente indeterminados (NaN). Además, sin tratar la no-estacionariedad, la red corre el riesgo de memorizar tendencias temporales espurias en lugar de la lógica térmica subyacente, lo que se traduce en un modelo con una precisión inflada en el entrenamiento que fallará catastróficamente ante cualquier cambio mínimo en el entorno real, resultando en un desperdicio crítico de tiempo y recursos de cómputo.
